```python
import yfinance as yf
tickers = ['RELIANCE.NS', 'HDFCBANK.NS', 'ICICIBANK.NS', 'INFY.NS', 'TCS.NS']
data = yf.download(tickers,start='2023-01-01',end='2026-01-01',auto_adjust=True)
```

```python
prices = data['Close']
returns = prices.pct_change() .dropna()
returns.head()
```

```python
icici_returns = returns['INFY.NS'].values
mean_return = np.mean(icici_returns)
volatility = np.std(icici_returns)
```

```python
all_means = np.mean(returns.values,axis=0)*252

all_std= np.std(returns.values,axis=0)*np.sqrt(252)
```

```python
annual_return = mean_return * 252
annual_volatility = volatility * np.sqrt(252)
```

```python
weights = np.array([0.20, 0.15, 0.15, 0.30, 0.20])
print(np.sum(weights)) 
```

```python
# Covariance Matrix
cov_matrix = returns.cov()
annual_cov_matrix = cov_matrix * 252
```

```python
cov_matrix
```

```python
# Portfolio Volatility Using Matrix Algebra 
portfolio_variance = weights.T @ annual_cov_matrix.values @ weights
portfolio_volatility = np.sqrt(portfolio_variance)
print( 'Portfolio Volatility:', portfolio_volatility)
```

```python
# Portfolio Return
annual_returns = returns.mean() * 252
portfolio_return = np.dot(weights, annual_returns.values)
print( 'Portfolio Return:', portfolio_return)
```

```python
#Sharpe Ratio 
risk_free_rate = 0.06
sharpe_ratio = (portfolio_return - risk_free_rate) / portfolio_volatility
print('Sharpe Ratio:', sharpe_ratio)
```

```python
num_portfolios = 10000

portfolio_returns = []
portfolio_volatility = []
sharpe_ratios = []
all_weights = []

risk_free_rate = 0.06

for i in range(num_portfolios):

    weights = np.random.random(len(tickers))
    weights = weights / np.sum(weights)

    portfolio_return = np.sum(weights * annual_returns)

    portfolio_vol = np.sqrt(
        np.dot(weights.T, np.dot(annual_cov_matrix, weights))
    )

    sharpe_ratio = (portfolio_return - risk_free_rate) / portfolio_vol

    portfolio_returns.append(portfolio_return)
    portfolio_volatility.append(portfolio_vol)
    sharpe_ratios.append(sharpe_ratio)
    all_weights.append(weights)
```

```python
max_sharpe_index = np.argmax(sharpe_ratios)

print("Maximum Sharpe Ratio:", sharpe_ratios[max_sharpe_index])
print("Portfolio Return:", portfolio_returns[max_sharpe_index])
print("Portfolio Volatility:", portfolio_volatility[max_sharpe_index])

print("\nOptimal Portfolio Weights:")

for ticker, weight in zip(tickers, all_weights[max_sharpe_index]):
    print(ticker, ":", round(weight * 100, 2), "%")
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    portfolio_volatility,
    portfolio_returns,
    c=sharpe_ratios
)

# Highlight portfolio with maximum Sharpe Ratio
plt.scatter(
    portfolio_volatility[max_sharpe_index],
    portfolio_returns[max_sharpe_index],
    marker="*",
    s=300,
    label="Maximum Sharpe Ratio"
)

plt.xlabel("Portfolio Volatility (Risk)")
plt.ylabel("Expected Portfolio Return")
plt.title("Monte Carlo Simulation: Portfolio Risk vs Return")

plt.colorbar(label="Sharpe Ratio")

plt.legend()

plt.show()
```
